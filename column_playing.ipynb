```python
import pandas as pd
```

```python
data = pd.read_excel("/content/sample_data/Data/Student_Marks_100.xlsx")
print(data.head(10))
```

```python
cust = pd.Series(data["Math"] >=80)
print(type(cust))
data.loc[(data["Math"] >=80 ),"pass/fail"] = True
data.loc[(data["Math"] < 80 ),"pass/fail"] = False
print(data)
```

```python
data["pass /fail"] =  data["Math"].map(lambda x : "pass" if x >= 80 else "fail")
print(data)
```

```python
print(data.iloc[1,2])
```

```python
data["Total"] = data["Math"]+data["English"]+data["Science"]
data["parcentage"] = (((data["Math"]+data["English"]+data["Science"])*100)/300).round(2)
data["parcentage"] = data["parcentage"].astype(str) +"%"
print(data)
```
